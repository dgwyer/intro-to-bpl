## Writing the formula for one item

Index notation lets you write a calculation as the formula for **one item** of the answer, using named positions called *indices*. For example, "item `i` of `v`, times 2" or "row `i`, column `j` of a table". BPL then runs that formula on the whole array at once. It's like Einstein notation, or NumPy's `einsum`.

We start with a single index and build up, one idea at a time, to matrix products. Two richer examples come at the end.

BPL docs: [Models in index notation](https://answerdotai.github.io/basedpl/index-notation.html)

Glyphs and ideas this dialog teaches:

| Glyph | Plain English |
|---|---|
| `•index n` | an index: the positions `0 1 … n-1`, with a hidden name of its own |
| `[i j]←•index 3 2` | make several indices at once |
| `•index ρm` | one index for each axis of `m` |
| `v.[i]` | the items of `v` at positions `i` |
| `+/⍠i` | add up along index `i` |
| `⁝` | keeps two things apart without ending the run: `+/⍠i⁝x` keeps `i` as the option, separate from `x` |

Other glyphs used along the way:

| Glyph | Plain English |
|---|---|
| `←` | store a value in a name |
| `⍝` | comment: BPL ignores the rest of the line |
| `ι3` | the positions `0 1 2` |
| `+ - × ÷` | add, subtract, multiply, divide |
| `+/` `⌈/` | total, biggest |
| `≡` | match: `$t` if two things are the same |
| `[… ◇ …]` | a table written row by row; `◇` starts the next row |
| `ρ` | the shape of an array (rows, columns) |
| `+.×` | BPL's built-in matrix product |
| `|` | size of a number, ignoring its sign |
| `,` | flatten a table into one list |
| `7○` | tanh, a smooth S-shaped curve |
| `⍤1` | rank: apply a function to each row |

### Indices

#### 1. An index

- `ι3` gives the positions `0 1 2`.
- `•index 3` gives the same positions, but its axis also gets a **hidden name**, unique to that index. The name is what lets BPL line things up later.
- The output `["_·…":3]ρ[0 1 2]` means "3 items along an axis with a hidden name". The numbers are `0 1 2`.

````
%%bpl
ι3               ⍝ plain positions
i←•index 3       ⍝ the same positions, with a hidden name
i
````

#### 2. Read a list with an index

- `v.[i]` reads the items of `v` at positions `i`. Here that's every item, in order.
- So `v.[i]×2` is the formula "item `i` of `v`, times 2", and BPL works it out for every `i` at once.
- `≡` checks it matches plain `v×2`.

````
%%bpl
v←10 20 40
v.[i]            ⍝ every item of v
v.[i]×2          ⍝ item i, times 2
(v.[i]×2)≡v×2    ⍝ $t: the same as plain v×2
````

#### 3. The same index pairs with itself

- When the same index appears twice, BPL pairs them: item `i` with item `i`.
- So `v.[i]+v.[i]` adds each item to itself.

````
%%bpl
v.[i]
v.[i]+v.[i]      ⍝ 20 40 80
````

### Grids

#### 4. Two different indices make a grid

- `[i j]←•index 3 3` makes two indices at once, each `0 1 2`, each with its own name.
- Because `i` and `j` have **different** names, BPL pairs every `i` with every `j`. The answer is a table, with one row for each `i` and one column for each `j`.
- `a.[i]×a.[j]` is a multiplication table.

````
%%bpl
a←1 2 3
[i j]←•index 3 3     ⍝ two separate indices
a.[i]×a.[j]          ⍝ row i, column j holds a[i] × a[j]
````

#### 5. The first index gives the rows

- With lists of different lengths, you can see which index goes where.
- The index that appears first (reading left to right) gives the rows, and the next one gives the columns.
- `a.[i]+b.[j]` has 3 rows (for `i`) and 2 columns (for `j`). Swap the order and you get 2 rows and 3 columns.

````
%%bpl
b←10 20
a
b

[i j]←•index 3 2     ⍝ i counts 3, j counts 2
a.[i]+b.[j]          ⍝ 3 rows, 2 columns
b.[j]+a.[i]          ⍝ 2 rows, 3 columns
````

#### 6. One index twice on a table: the diagonal

- `s.[k k]` uses the same index for the row **and** the column, so it reads `s` at row 0 column 0, row 1 column 1, and so on.
- That's the diagonal. Adding it up gives 1+5+9.

````
%%bpl
s←[1 2 3◇4 5 6◇7 8 9]
k←•index 3
s
k

s.[k k]          ⍝ the diagonal: 1 5 9
+/s.[k k]        ⍝ 15
````

### Sums

#### 7. Add up along one index

Start with the table `t` from step 5:

```text
11 21
12 22
13 23
```

The table has two named directions:

- `i` identifies the **rows**. There are three rows, one for each value of `a`.
- `j` identifies the **columns**. There are two columns, one for each value of `b`.

Now read `+/⍠j⁝t` one piece at a time:

- `+/` means **add several numbers together**.
- `⍠j` tells BPL **which named direction to add along**: the `j` direction.
- `⁝` is only a separator. It marks the end of the option `j` and the beginning of the data `t`.
- `t` is the table whose numbers we want to add.

So `+/⍠j⁝t` means: **in each row of `t`, add the numbers across the `j` columns**.

- First row: `11+21` gives `32`.
- Second row: `12+22` gives `34`.
- Third row: `13+23` gives `36`.

The result is therefore `32 34 36`. The `j` direction has been added away, while the `i` direction remains, so there is one result for each row.

Changing `j` to `i` changes the direction of the sum. `+/⍠i⁝t` means: **for each column, add down the `i` rows**.

- First column: `11+12+13` gives `36`.
- Second column: `21+22+23` gives `66`.

The result is `36 66`. This time the `i` direction has been added away, while the `j` direction remains, so there is one result for each column.

````
%%bpl
t←a.[i]+b.[j]    ⍝ the 3×2 table from step 5
+/⍠j⁝t           ⍝ row totals: add along j
+/⍠i⁝t           ⍝ column totals: add along i
````

#### 8. Dot product

- Multiply two lists item by item, then add up: "sum over `i` of `a[i] × c[i]`".
- In index notation that's `+/⍠i⁝a.[i]×c.[i]`. It's the same as plain `+/a×c`.

````
%%bpl
a←1 2 3
c←4 5 6
a
c
i←•index 3
+/⍠i⁝a.[i]×c.[i]     ⍝ 1×4 + 2×5 + 3×6
+/a×c                ⍝ the same
````

#### 9. A table times a list

- `[r c]←•index ρm` makes one index for each axis of `m`: `r` for its rows and `c` for its columns.
- The formula for answer item `r` is "sum over `c` of `m[r c] × v[c]`".
- `+.×` is BPL's built-in matrix product. It gives the same numbers.

````
%%bpl
m←[1 2 3◇4 5 6]      ⍝ 2 rows, 3 columns
v←10 20 30
m
v

[r c]←•index ρm      ⍝ one index per axis
r
c
m.[r c] v.[c]
+/⍠c⁝m.[r c]×v.[c]   ⍝ sum over c
m+.×v                ⍝ the same
````

#### 10. Matrix product

- Two tables: `x` is 2×3 and `w` is 3×2. Answer item `[b o]` is "sum over `k` of `x[b k] × w[k o]`".
- `k` appears in both, so those line up. `b` and `o` are different, so they make the grid of the answer.
- BPL spots this pattern and runs it as a real matrix product, so the formula is as fast as `x+.×w`.

````
%%bpl
x←[1 2 3◇4 5 6]          ⍝ 2 rows, 3 columns
w←[1 0◇0 1◇1 1]          ⍝ 3 rows, 2 columns
[b k o]←•index 2 3 2     ⍝ rows of x, the shared k, columns of w
y←+/⍠k⁝x.[b k]×w.[k o]
y
y≡x+.×w                  ⍝ $t
````

### Two richer examples

#### 11. The biggest change between any two days

- A week of temperatures. `temps.[i]-temps.[j]` is a 7×7 table of every day minus every other day.
- `|` drops the sign, `,` flattens the table into one list, and `⌈/` takes the biggest: 20 − 9 = 11.

````
%%bpl
temps←12 15 11 18 20 14 9
[i j]←•index 7 7
temps 
i
j
d←|temps.[i]-temps.[j]   ⍝ every pair of days
temps.[i]-temps.[j]
d
,d
⌈/,d                     ⍝ the biggest change
````

#### 12. One layer of a neural network

A layer takes inputs `x` (one row per sample), weights `w` and a bias, and gives one output per sample and hidden unit.

- Make the indices from the shapes: `n` for samples, `i` for inputs (both from `ρx`), and `h` for hidden units (the length of `bias`).
- The formula for one output: "tanh of (sum over `i` of `x[n i] × w[i h]`) plus `bias[h]`". `7○` is tanh.
- The usual tacit form is `7○bias(+⍤1)x+.×w`, where `⍤1` adds the bias to each row. Both give the same answer.

````
%%bpl
x←[1 2 3◇4 5 6]                  ⍝ 2 samples, 3 inputs
w←[0.1 ¯0.1◇0.2 0◇¯0.1 0.1]      ⍝ 3 inputs, 2 hidden units
bias←0.5 ¯0.5
[[n i] h]←•index[ρx ≢bias]       ⍝ n, i from ρx; h from the bias length
a←7○(+/⍠i⁝x.[n i]×w.[i h])+bias.[h]
a
a≡7○bias(+⍤1)x+.×w               ⍝ $t: same as the tacit form
````
