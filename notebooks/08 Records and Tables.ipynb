## Records and tables

So far our data has been plain lists. In this dialog we give data **names**: first a single *record* (like one row on a form), then a whole *table* made of named columns. Then we filter, sort, add columns and group, much as you would in a spreadsheet.

Glyphs and ideas this dialog teaches:

| Glyph | Plain English |
|---|---|
| `["key":value]` | a record: values with names. `:` joins each name to its value |
| `.name` | read the field called `name` |
| `+←` | change a stored value: `x+←1` adds 1 to `x` |
| `¨` | each: do a function to every item (here, every column) |
| `.[ ]` | pick items by position or by name: `v.[i]` |
| `keys:values` | put names on the items of a list |
| `⌸` | key: group items that share a label |
| `∪` | unique: each different item once |
| `•csv` | read CSV text into a table |

Other glyphs used along the way:

| Glyph | Plain English |
|---|---|
| `←` | store a value or function in a name |
| `⍝` | comment: BPL ignores the rest of the line |
| `"…"` | text |
| `{ }` and `ω` | a function written with braces; `ω` is its right argument |
| `+ × ÷` | add, multiply, divide |
| `>` | greater than (answers `$t` true or `$f` false) |
| `+/` | total of a list |
| `≢` | count the items |
| `+/÷≢` | a train: total ÷ count, which is the mean |
| `⊢` | give back the right argument |
| `#` | keep the items where the answer is `$t` |
| `⍋` `⍒` | the positions that would sort a list, smallest first / biggest first |
| `,` | join two pieces of text |
| `•ucs 10` | the newline character |
| `◇` | separates two statements on one line |
| `ₓ` | (in output) marks an exact number |

### The data

A week of midday temperatures (°C) and rainfall (mm), Monday to Sunday.

```
%%bpl
days←"Mon" "Tue" "Wed" "Thu" "Fri" "Sat" "Sun"
temps←12 15 11 18 20 14 9     ⍝ °C
rain←0 2 5 0 0 1 8            ⍝ mm
days
temps
rain
```

### Records

#### 1. A record: values with names

- A record holds several values, each with a name (called a *key*).
- Write it in square brackets: `["name":"Rex" "age":3]`.
- `:` joins a name (left) to its value (right). Spaces separate the pairs.
- `.name` after a record reads one field.

```
%%bpl
pet←["name":"Rex" "age":3]   ⍝ a record with two fields
pet.name                      ⍝ "Rex"
pet.age                       ⍝ 3
```

#### 2. Change a field, or add a new one

- `pet.age+←1` means "add 1 to `pet.age`". `+←` changes the stored value in place.
- Storing into a field that doesn't exist yet adds it: `pet.legs←4`.

```
%%bpl
pet.age+←1      ⍝ a birthday: age 3 becomes 4
pet.legs←4      ⍝ a new field
pet
```

### Tables

#### 3. A table is a record of columns

- Give each column a name, and put them in one record.
- Every column has the same length, so position 0 of each column is Monday, position 1 is Tuesday, and so on. One position across all columns is one **row**.

```
%%bpl
T←["day":days "temp":temps "rain":rain]
T
```

The output shows each column in its own box, side by side. To read one row, take the same position from every box: the first row is Mon, 12°C, 0 mm.

```
%%bpl
monday←["day":T.day.[0] "temp":T.temp.[0] "rain":T.rain.[0]]
monday
```

#### 4. One column

- `T.temp` reads the `temp` column. It's an ordinary list, so the usual glyphs work on it.
- `+/T.rain` adds up the rain: 16 mm this week.

```
%%bpl
T.temp        ⍝ the temperature column
T.rain        ⍝ the rain column
+/T.rain      ⍝ total rain
```

#### 5. Several columns

- Put the column names in a name first: `nums←"temp" "rain"`.
- `T.[nums]` picks those columns, and gives back a smaller table.
- Why a name? A list written straight inside `.[ ]` is read as one choice per axis, which isn't what we want here.

```
%%bpl
nums←"temp" "rain"   ⍝ the number columns
N←T.[nums]           ⍝ a table of just those two
N
```

#### 6. Every column at once: `¨`

- `¨` (each) does a function to every item. For a table, every item is a column.
- `≢¨T` counts each column: 7 days in every one.
- `mean←+/÷≢` is a train: total ÷ count. `mean¨N` gives the mean of each number column.
- The answers keep the column names.

```
%%bpl
N
≢¨T              ⍝ length of each column
mean←+/÷≢        ⍝ total ÷ count
mean¨N           ⍝ mean temperature and mean rain
```

#### 7. Keep some rows

- `warm←T.temp>14` gives `$t` (true) for the warm days.
- `#` keeps the items where the answer is `$t`: `warm#T.day` gives the warm days' names.
- To keep whole rows, do that to every column with `¨`: `{warm#ω}¨T`. Inside the braces, `ω` is one column.

```
%%bpl
warm←T.temp>14    ⍝ which days are above 14°C?
warm#T.day        ⍝ their names
{warm#ω}¨T        ⍝ whole rows: every column filtered the same way
```

#### 8. Sort the rows

- `⍋T.temp` gives the positions that would sort the temperatures, coldest first. Store them in `i`.
- `v.[i]` reads `v` at those positions, so `T.day.[i]` lists the days coldest first.
- `{ω.[i]}¨T` reorders every column the same way, so each row stays together.

```
%%bpl
T.temp
⍋T.temp
i←⍋T.temp         ⍝ positions, coldest first
T.day.[i]         ⍝ days, coldest first
{ω.[i]}¨T         ⍝ the whole table, sorted by temperature
```

#### 9. Add a column

- Store into a new field, as in step 2. Here we add the temperature in Fahrenheit: times 1.8, then add 32.

```
%%bpl
T.F←32+1.8×T.temp   ⍝ new column F
T.day
T.F
```

### Labels and groups

#### 10. Names on the items of a list

- `days:temps` pairs the two lists **position by position**. Each item in `days` becomes a field name (a key), and the temperature at the same position becomes that field's value.
- The first few pairs are therefore `"Mon":12`, `"Tue":15`, and `"Wed":11`. Altogether, `W` is equivalent to a record like `["Mon":12 "Tue":15 ... "Sun":9]`.
- This works because `days` and `temps` have the same number of items. If the lists did not line up, it would not be clear which value belonged to which name.
- After storing that record in `W`, dot notation reads a value by name: `W.Tue` means "the field named `Tue`", so it returns `15`.
- Compare this with `temps.[1]`: both return Tuesday's temperature, but `W.Tue` uses a meaningful name instead of remembering a position.

```
%%bpl
W←days:temps    ⍝ each temperature labelled with its day
W.Tue           ⍝ 15
```

#### 11. Group rows: `⌸`

Suppose we want one average for weekdays and another for the weekend.

- `daytype` gives every temperature a label: the first five are `"weekday"`; the last two are `"weekend"`.
- `⌸` is **key**: it groups the values on its right using the matching labels on its left. It then applies the function immediately before `⌸` (here `(mean⊢)`) once to each group.
- In `daytype(mean⊢)⌸T.temp`, it collects temperatures with the same `daytype`, then finds each group's mean. `⊢` says to use the temperatures, not the labels.
- `∪daytype` keeps each different label once, in first-seen order. `labels:avgtemp` attaches those labels to the two answers.

```
%%bpl
daytype←"weekday" "weekday" "weekday" "weekday" "weekday" "weekend" "weekend"
avgtemp←daytype(mean⊢)⌸T.temp   ⍝ one mean for each label
labels←∪daytype                 ⍝ each different label once
labels:avgtemp                  ⍝ attach labels to the answers
```

**Walkthrough**

1. `daytype` has one label for each item in `T.temp`. The first five temperatures are labelled `"weekday"`; the last two are labelled `"weekend"`.
2. In `daytype(mean⊢)⌸T.temp`, `daytype` is on the **left** of `⌸` and supplies the labels; `T.temp` is on the **right** and supplies the values. `⌸` matches them by position:
   - the first label, `"weekday"`, matches the first temperature, `12`
   - the second `"weekday"` matches `15`, and so on
   It gathers right-side values whose left-side labels are equal, forming:
   - `"weekday"` → `12 15 11 18 20`
   - `"weekend"` → `14 9`
   It then calls `(mean⊢)` once for each group. `⊢` selects that group's temperatures (the right argument), and `mean` averages them. Thus `avgtemp` is `15.2 11.5`.
3. `∪daytype` scans the labels from left to right. It keeps the first `"weekday"` and discards its later repeats; then it keeps the first `"weekend"` and discards its repeat. This produces `"weekday" "weekend"`—one label per group. `⌸` returns its group results in the same first-seen order, so these labels line up with `15.2 11.5`.
4. `labels:avgtemp` pairs each label with its average, producing `["weekday":15.2 "weekend":11.5]`.

#### 12. Read a table from CSV

- CSV is text with one row per line and commas between the values. The first line holds the column names.
- `•ucs 10` is the newline character, stored in `nl`. `,` joins pieces of text.
- `•csv` turns the text into a table just like `T`, a record of columns.

```
%%bpl
nl←•ucs 10
csv←"day,temp,rain",nl,"Mon,12,0",nl,"Tue,15,2",nl,"Wed,11,5"
csv
S←•csv csv        ⍝ the CSV text as a table
S.temp            ⍝ its temp column
```
