## Text and strings

In BPL, a piece of text is just a list of characters. That means almost everything from our BPL lists dialog (counting, comparing, filtering, reversing, grouping) works on text too. This dialog shows how, and adds a few text-specific tools.

Glyphs and functions in this dialog:

| Glyph | Plain English |
|---|---|
| `'a'` `"abc"` | one character / a piece of text |
| `,` | join two lists together |
| `=` `≠` | equal / not equal |
| `⍷` | find where a pattern starts |
| `•c` | change upper / lower case |
| `⊆` | split a list into pieces |
| `¨` | apply a function to each item |
| `⍕` `⍎` | number to text / text to number |
| `•r` | regular expressions: find every match of a pattern |
| `•replace` | swap every match of a pattern for new text |

#### 1. One character vs a piece of text

BPL uses two kinds of quotes:

- **Single quotes** make **one character**: `'a'`.
- **Double quotes** make **text**, which is a list of characters: `"abc"`.

`ρ` (shape) shows the difference. A single character has no length at all, so its shape is empty: `⍬` means "an empty list". Text, even one letter long, is a list, so its shape is its length.

```
%%bpl
ρ'a'      ⍝ one character: no length
⍝ 'ab'   not allowed, only single characters
ρ"a"      ⍝ text with one character in it: length 1
ρ"a "     ⍝ text with two characters (spaces count too!): length 2
ρ"abc"    ⍝ text with three characters: length 3
```

Because text is a list of characters, writing two characters side by side also makes text: `'h' 'i'` is the same as `"hi"`.

```
%%bpl
'h' 'i'
('h' 'i')≡"hi"   ⍝ ≡ (match): are they the same?
```

#### 2. Length: `≢`

`≢` counts the items in a list. For text, the items are characters, so it counts letters (spaces count too).

```
%%bpl
≢"hello"
≢"hello world"   ⍝ 11: the space is a character too
```

#### 3. Joining text: `,`

`,` (*catenate*) joins two lists into one. For text, that glues pieces together.

```
%%bpl
"Hello",", ","world"
```

BPL reads from right to left, so this joins `", "` and `"world"` first, then puts `"Hello"` in front. For joining, the result is the same either way.

#### 4. Compare every letter: `=`

`s='a'` asks "is this letter an a?" for **every** character of `s` at once, giving one true/false per letter. `+/` counts the trues and `⍸` gives their positions (counting from 0).

```
%%bpl
s←"banana"
s='a'      ⍝ is each letter an 'a'?
+/s='a'    ⍝ how many a's
⍸s='a'     ⍝ where they are
```

#### 5. Count the vowels: `∊`

`a∊b` asks "is each item of `a` somewhere in `b`?". With text on both sides, `s∊"aeiou"` asks "is each letter a vowel?".

```
%%bpl
s∊"aeiou"      ⍝ is each letter a vowel?
+/s∊"aeiou"    ⍝ how many vowels
```

#### 6. Find a pattern: `⍷`

`=` checks one letter at a time. To look for a **group** of letters, use `⍷` (*find*). `"an"⍷s` marks every position where `"an"` **starts** in `s`.

```
%%bpl
"an"⍷s     ⍝ true where "an" starts
⍸"an"⍷s    ⍝ the starting positions
```

"an" starts at positions 1 and 3 of "banana" (b is position 0).

#### 7. Upper and lower case: `•c`

Names starting with `•` are **system functions**: built-in tools with a name rather than a single glyph. `•c` changes case:

- `•c text` makes it **lower** case.
- `1•c text` makes it **upper** case (the 1 on the left means "up").

```
%%bpl
•c "HeLLo"     ⍝ lower case
1•c "hello"    ⍝ upper case
```

#### 8. Reverse, and spot a palindrome: `⌽` `≡`

`⌽` reverses a list, so on text it spells the word backwards. A **palindrome** reads the same both ways, so a word is a palindrome when it matches (`≡`) its own reverse.

```
%%bpl
⌽"stressed"          ⍝ backwards
w←"level"
w≡⌽w                 ⍝ the same backwards? yes
w←"hello"
w≡⌽w                 ⍝ no
```

#### 9. Split a sentence into words: `≠` `⊆`

Two steps:

1. `s≠' '` asks "is this character **not** a space?" for every character. (`≠` means *not equal*.)
2. `⊆` (*partition*) takes those true/false marks on the left and the text on the right. It keeps each unbroken run of trues as one piece and throws away the falses (the spaces).

Small example first: the marks `1 1 0 1 1` keep "ab", drop the space, and keep "de".

```
%%bpl
1 1 0 1 1⊆"ab de"   ⍝ keep, keep, drop, keep, keep
```

Now the same with a real sentence.

```
%%bpl
s←"the cat sat on the mat"
s≠' '                ⍝ true for every letter, false for spaces
words←(s≠' ')⊆s      ⍝ split into words
words
≢words               ⍝ how many words
```

The result is a list of 6 pieces of text, one per word. (Each word prints in its own box.)

#### 10. Do something to each word: `¨`

`≢words` counts the **words**. To count the letters **in each word**, add `¨` (*each*): `≢¨words` applies `≢` to every word separately.

```
%%bpl
≢ words      ⍝ number of words
≢¨words      ⍝ letters in each word
```

#### 11. Count repeated words: `⌸`

`⌸` groups equal items and calls its function once for each different item. The function’s right argument (`ω`) contains that item’s positions, so counting it tells us how often the item appears.

Below we show three equivalent versions:

- `{≢ω}` is an **explicit** function: count the right argument.
- `fn←≢⊢` defines the same function **tacitly** (without naming `ω`): `⊢` selects the right argument, then `≢` counts it.
- `(≢⊢)⌸words` uses that tacit function directly, without first giving it a name.

The different words remain in the same order as `∪words`.

```
%%bpl
[∪words ◇ {≢ω}⌸words]   ⍝ each different word, with its count underneath
fn←≢⊢
[∪words ◇ fn⌸words]     ⍝ tacit fn version
[∪words ◇ (≢⊢)⌸words]   ⍝ tacit direct version
```

"the" appears twice; every other word appears once.

#### 12. Numbers to text and back: `⍕` `⍎`

You can't join a number straight onto text, because a number isn't a list of characters. `⍕` (*format*) turns a number into text first. Then `,` can join it.

```
%%bpl
⍕42               ⍝ the number 42 as text
≢⍕42              ⍝ it is 2 characters long
"Total: ",⍕+/1 2 3
```

A number on the left of `⍕` sets how many decimal places to show.

```
%%bpl
temps←12 15 11 18 20 14 9
avg←(+/temps)÷≢temps
"Average: ",⍕avg       ⍝ every digit
"Average: ",1⍕avg      ⍝ 1 decimal place
```

One detail: counts from `≢` are **exact** numbers, and `⍕` writes exact numbers with their `ₓ` mark, which looks odd in a sentence. Putting `⍬` (an empty list) on the left, `⍬⍕`, leaves the mark out.

```
%%bpl
⍕≢words                             ⍝ includes the ₓ mark
⍬⍕≢words                            ⍝ without it
"There are ",(⍬⍕≢words)," words"    ⍝ without the mark
"There are ",(⍕≢words)," words"     ⍝ with the mark
```

Going the other way, `⍎` (*execute*) reads text as BPL code and runs it. On text that holds a number, that turns it back into a number you can do arithmetic with.

```
%%bpl
⍎"42"       ⍝ text → number
1+⍎"42"     ⍝ now it can be used in sums
```

#### 13. Ranges of letters: `↦`

You met `↦` (range) with numbers: `3↦9` counts from 3 to 9. It works with characters too, counting through the alphabet.

```
%%bpl
'a'↦'e'
'a'↦'z'
```

#### 14. Pattern matching: `•r`

A **regular expression** (regex) is a short pattern that describes text to look for. For example:

- `[0-9]` means "any digit".
- `+` after it means "one or more of those".

So `"[0-9]+"` means "a run of digits", in other words a whole number.

- `text •r pattern` finds every match of `pattern` in `text`. It gives a small table: `.text` lists the matched pieces, and `.position` says where each one starts.
- `text •replace [pattern new]` swaps every match for `new`. The pattern and the new text go together in square brackets.

```
%%bpl
s←"I have 3 cats and 12 fish"
m←s •r "[0-9]+"            ⍝ find every run of digits
m.text                     ⍝ the matched pieces: "3" "12"
m.position                 ⍝ where each one starts
```

```
%%bpl
"my cat likes your cat" •replace ["cat" "dog"]   ⍝ swap every cat for dog
```
