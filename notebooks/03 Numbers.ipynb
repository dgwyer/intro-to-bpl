## Exact, approximate, and a few special numbers

You've seen a small `ₓ` after many answers, like `6ₓ`. This dialog explains it. BPL has three everyday kinds of number: **approximate** decimals, for measurements; **whole numbers** (`ₓ`), for counting; and **exact** numbers of any size, including fractions (`ᵣ`). This dialog is written for BPL 0.1.85, which changed how `ₓ` and `ᵣ` work. It also has a few special values: infinity, true and false, and complex numbers.

BPL docs: [Numbers](https://answerdotai.github.io/basedpl/numbers.html)

Glyphs and ideas this dialog teaches:

| Glyph | Plain English |
|---|---|
| `3`, `0.5` | an approximate number (a decimal, also called a float) |
| `3ₓ` | a whole number (an ordinary computer integer). `ₓ` in output marks whole numbers |
| `3ᵣ`, `1r3` | an exact number of any size. `1r3` is exactly one third, shown as `1ᵣ3` |
| `÷ₐ` `÷ᵣ` | divide, giving a decimal / an exact fraction |
| `¯2` | a negative number. `¯` is part of the number |
| `1E6` | a number with an exponent: 1 × 10⁶. Output writes it as `1ₑ6` |
| `*` | power: `2*3` is 8 |
| `!` | factorial: `!5` is 5×4×3×2×1 |
| `√` | square root |
| `⌊` `⌈` | round down / round up |
| `∞` | infinity |
| `$t` `$f` | true and false, which count as 1 and 0 |
| `2j3` | a complex number, 2 + 3i. It's shown as `2ⱼ3` |
| <code>&#124;</code> | size of a number, ignoring its sign |
| `⍕` | turn a number into text. `2⍕x` rounds to 2 decimal places |
| `⍎` | turn text into a number by running it |

Other glyphs used along the way:

| Glyph | Plain English |
|---|---|
| `←` | store a value in a name |
| `⍝` | comment: BPL ignores the rest of the line |
| `+ - × ÷` | add, subtract, multiply, divide |
| `=` `>` | compare (answers `$t` or `$f`) |
| `≢` `ρ` | count, shape |
| `+/` `⌈/` | total, biggest |
| `⍬` | the empty list |

### Three kinds of number

#### 1. Plain numbers are approximate

- A number you type, like `1` or `0.5`, is **approximate**: a decimal stored to about 16 digits.
- `1÷3` is 0.333…, cut off at some point. The display shows 10 digits.

```
%%bpl
1÷3         ⍝ approximate: 0.3333333333
2÷4         ⍝ 0.5
```

#### 2. Whole numbers: `ₓ`

- Put `ₓ` after a whole number to make it a BPL **whole number**: an ordinary computer integer, such as `3ₓ`.
- Adding, subtracting and multiplying whole numbers gives whole numbers.
- **Dividing** whole numbers also gives a whole number, rounded **down**: `7ₓ÷2ₓ` is `3ₓ`, and `1ₓ÷3ₓ` is `0ₓ`.
- To get a decimal instead, put `ₐ` after the `÷`: `7ₓ÷ₐ2ₓ` is 3.5.

```
%%bpl
3ₓ+4ₓ        ⍝ 7ₓ
7ₓ÷2ₓ        ⍝ 3ₓ: rounded down
1ₓ÷3ₓ        ⍝ 0ₓ: rounded down
7ₓ÷ₐ2ₓ       ⍝ 3.5: ÷ₐ gives a decimal
```

#### Exact fractions: `ᵣ`

- `ᵣ` marks an **exact** number of any size. Exact numbers divide into exact fractions: `1ᵣ÷3ᵣ` is exactly one third, shown as `1ᵣ3`.
- You can type a fraction directly as `1r3`. Fractions add exactly: a third plus a sixth is exactly a half.
- `÷ᵣ` divides two whole numbers into an exact fraction: `7ₓ÷ᵣ2ₓ` is `7ᵣ2`.
- A whole answer stays exact, and shows a trailing `ᵣ`: three thirds make `1ᵣ`.

```
%%bpl
1ᵣ÷3ᵣ        ⍝ exactly one third: 1ᵣ3
1r3+1r6      ⍝ exactly one half
7ₓ÷ᵣ2ₓ       ⍝ ÷ᵣ gives an exact fraction: 7ᵣ2
1r3+1r3+1r3  ⍝ exactly 1, shown as 1ᵣ
```

#### 3. Why `ₓ` keeps appearing

- Counting always gives a whole number, so counts are `ₓ` whole numbers. That's why `≢` and `ρ` answer with `ₓ`.
- Rounding with `⌊` and `⌈` gives `ₓ` whole numbers too.

```
%%bpl
≢10 20 30     ⍝ a count: 3ₓ
2 3ρ0
ρ2 3ρ0        ⍝ a shape: [2 3]ₓ
⌊2.7          ⍝ rounded down: 2ₓ
```

#### 4. Mixing exact numbers and decimals

- Exact and approximate together give an **approximate** answer, because the approximate part can't be made exact.
- So `1r2+0.5` is `1`, with no `ᵣ` or `ₓ`.

```
%%bpl
1r2+0.5       ⍝ approximate 1
3ₓ×0.5        ⍝ approximate 1.5
```

#### 5. Very big numbers

- `*` is power: `2*3` is 8. `!` is factorial: `!5` is 5×4×3×2×1.
- `ₓ` whole numbers have room for about 19 digits. Past that they **wrap around**, like a car's odometer, and give a wrong answer with no warning: `2ₓ*64ₓ` is `0ₓ`.
- `ᵣ` numbers grow as big as needed and keep **every digit**: `2ᵣ*64ₓ`.
- Approximate numbers keep about 16 digits, and show very big values with an exponent: `1.84…ₑ19` means 1.84… × 10¹⁹.

```
%%bpl
2ₓ*62ₓ        ⍝ fits
2ₓ*64ₓ        ⍝ too big: wraps round to 0ₓ
2ᵣ*64ₓ        ⍝ ᵣ: every digit
2*64          ⍝ approximate: about 1.84 × 10¹⁹
!25ᵣ          ⍝ 25 factorial, exactly
```

### Writing numbers

#### 6. Negative numbers: `¯` is not `-`

- `¯3` is the number minus three. `¯` (high minus: press **Alt-Minus**, then Space) is part of the number.
- `-` is a **function**. BPL works right to left, so `-3+4` means `-(3+4)`, which is ¯7.

```
%%bpl
¯3+4          ⍝ minus three, plus four: 1
-3+4          ⍝ negate (3+4): ¯7
```

#### 7. Exponents

- `E` writes "times 10 to the power": `1E6` is a million and `1E¯3` is a thousandth.

```
%%bpl
1E6           ⍝ 1000000
1E¯3          ⍝ 0.001
1.5E3         ⍝ 1500
```

### Decimals are close, not exact

#### 8. A tiny difference

- Computers store decimals in binary, so `0.1` and `0.2` are very slightly off.
- `0.1+0.2` *displays* as 0.3, but subtracting 0.3 leaves a tiny difference, and `⍕` (all digits) shows it.
- BPL's `=` allows for this tiny difference, so `(0.1+0.2)=0.3` is `$t`.
- The brackets matter: without them, BPL reads right to left and compares `0.2=0.3` first.

```
%%bpl
0.1+0.2           ⍝ shows 0.3
(0.1+0.2)-0.3     ⍝ a tiny difference: about 5.6 × 10⁻¹⁷
⍕0.1+0.2          ⍝ every digit, as text
(0.1+0.2)=0.3     ⍝ $t: = allows for it
```

#### 9. Rounding

- `⌊` rounds down and `⌈` rounds up.
- To round to the **nearest** whole number, add 0.5 and round down: `⌊0.5+x`.
- To show a number to a few decimal places, `2⍕x` gives text rounded to 2 places.

```
%%bpl
x←7÷3
x             ⍝ 2.333333333
⌊x            ⍝ round down: 2
⌈x            ⍝ round up: 3
⌊0.5+x        ⍝ nearest: 2
2⍕x           ⍝ "2.33", as text
```

### Special values

#### 10. Infinity: `∞`

- Dividing by zero gives infinity (`∞`) rather than an error.
- `⌈/⍬` (the biggest of nothing) is `¯∞`. That's a sensible starting point: anything you compare with it is bigger.

```
%%bpl
1÷0           ⍝ ∞
¯1÷0          ⍝ ¯∞
⌈/⍬           ⍝ biggest of an empty list: ¯∞
```

#### 11. True and false are numbers

- `$t` is 1 and `$f` is 0, so you can add them up.
- That's why `+/temps>14` counts the warm days: it adds up the `$t`s.

```
%%bpl
$t+$t                    ⍝ 2
temps←12 15 11 18 20 14 9
temps
+/temps>14               ⍝ 3 warm days
```

#### 12. Complex numbers

- `2j3` is the complex number 2 + 3i (i is the square root of −1). It's shown as `2ⱼ3`.
- So `√¯4` doesn't fail: it gives `0ⱼ2`, which is 2i.
- `|` gives the size of a number. For 3 + 4i that's 5 (a 3-4-5 triangle).

```
%%bpl
2j3+1j1       ⍝ 3ⱼ4
√¯4           ⍝ 0ⱼ2: the square root of −4
|3j4          ⍝ size: 5
```

### Numbers and text

#### 13. Numbers to text and back: `⍕` and `⍎`

- `⍕` turns a number into text. It keeps the `ₓ` mark on whole numbers, and `⍬⍕` leaves the mark out (useful in sentences).
- `⍎` runs a piece of text as BPL, so `⍎"3.5"` gives the number 3.5.

```
%%bpl
⍕6ₓ               ⍝ "6ₓ"
⍬⍕6ₓ              ⍝ "6"
(⍎"3.5")+1        ⍝ text to number, then add 1: 4.5
```

#### 14. Fractions simplify themselves

- Exact fractions are automatically reduced to their simplest form.
- For example, `2r4` becomes one half, and `6r8` becomes three quarters.

```
%%bpl
2r4           ⍝ simplifies to one half
6r8           ⍝ simplifies to three quarters
```
