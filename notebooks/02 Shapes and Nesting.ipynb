## The shape of an array

Every BPL value is an *array*: a single number, a list, a table, or something with more sides. This dialog is about the **shape** of arrays: how long they are, how to arrange items into a table and back, how to join and flip them, and how a list can hold other lists (*nesting*).

BPL docs: [Arrays](https://answerdotai.github.io/basedpl/arrays.html)

Glyphs and ideas this dialog teaches:

| Glyph | Plain English |
|---|---|
| `ρx` | shape: the length of each side of `x` |
| `2 3ρx` | reshape: arrange the items of `x` into 2 rows and 3 columns |
| `≢ρx` | rank: how many sides (axes) `x` has |
| `,x` | ravel: flatten `x` into one list |
| `a,b` | join side by side (along the last axis) |
| `a⍪b` | join one on top of the other (along the first axis) |
| `⍉` | transpose: swap rows and columns |
| `[[1 2] [3 4 5]]` | a nested list: a list whose items are lists |
| `¨` | each: do a function to every item |
| `≡x` | depth: how many levels of nesting `x` has |
| `⊂` | enclose: wrap an array up as one item |
| `n⊃x` | pick: item `n` of `x`, unwrapped |
| `⊃` | mix: turn a list of equal-length lists into a table |

Other glyphs used along the way:

| Glyph | Plain English |
|---|---|
| `←` | store a value in a name |
| `⍝` | comment: BPL ignores the rest of the line |
| `≢` | count the items (along the first axis) |
| `↑` `↓` | take / drop items (or rows) from the start; a negative number counts from the end |
| `⌽` | reverse |
| `+/` | total of a list |
| `↦12` | the numbers 1 to 12 |
| `"…"` | text, which is a list of characters |
| `≡` (two arguments) | match: `$t` if two things are the same |
| `ₓ` | (in output) marks an exact number |

### Shape

#### 1. The shape of a list: `ρ`

- `≢v` counts the items: 6.
- `ρv` gives the **shape**: a list with one length for each side. A plain list has one side, so its shape is `[6]`, a one-item list holding 6.

```
%%bpl
v←10 20 30 40 50 60
≢v        ⍝ count: 6
ρv        ⍝ shape: [6], one side of length 6
```

```
%%bpl
u←[10 20 30 ◇ 40 50 60]
u
≢,u       ⍝ we have to revel first but the count is the same: 6
ρu        ⍝ shape: [6], one side of length 6
```

#### 2. Make a table: reshape

- With a shape on the left, `ρ` *reshapes*: `2 3ρv` arranges the items of `v` into 2 rows and 3 columns, row by row.
- `ρm` is now `2 3`: two sides.
- `≢m` counts the **rows** (the first side). `≢ρm` counts the sides themselves: a table has 2. This is called the *rank*.

```
%%bpl
m←2 3ρv       ⍝ 2 rows, 3 columns
m
ρm            ⍝ 2 3
≢m            ⍝ 2 rows
≢ρm           ⍝ 2 sides: rank 2
```

#### 3. Reshape repeats, and works on text

- If there aren't enough items, reshape goes back to the start and repeats them.
- `5ρ0` is five zeros, a quick way to start a list.
- Text is a list of characters, so it can be reshaped too.

```
%%bpl
3 4ρ1 2 3       ⍝ 1 2 3 repeated to fill 3×4
5ρ0             ⍝ five zeros
2 2ρ"abcd"      ⍝ a 2×2 table of letters
```

#### 4. Ravel (flatten): `,`

- `,m` (comma with nothing on the left) **ravels** a table: it flattens the table back into one list, row by row.

```
%%bpl
,m              ⍝ back to 10 20 30 40 50 60
```

### Joining and flipping

#### 5. Join: `,` and `⍪`

- With two arguments, `,` joins: `v,70` adds 70 to the end, `0,v` adds 0 to the start.
- On tables, `m,m` joins **side by side** (more columns), and `m⍪m` joins **one on top of the other** (more rows).

```
%%bpl
v,70            ⍝ add to the end
0,v             ⍝ add to the start
m,m             ⍝ side by side: 2 rows, 6 columns
m⍪m             ⍝ on top: 4 rows, 3 columns
```

Both the last two lines use a **dyadic** (two-argument) join: the first `m` is the left argument, the glyph is the function, and the second `m` is the right argument.

- `m,m` is parsed as `m , m`. Dyadic `,` joins along the **last axis** (columns), so corresponding rows are placed side by side. Two `2×3` tables become one `2×6` table.
- `m⍪m` is parsed as `m ⍪ m`. `⍪` joins along the **first axis** (rows), so the second table is placed below the first. Two `2×3` tables become one `4×3` table.

The unchanged axis must match: rows for `,`, columns for `⍪`.

#### 6. Take and drop rows

- You already know `↑` and `↓` on lists. On a table they work on **rows**.
- `1↑m` keeps the first row, and `1↓m` drops it.

```
%%bpl
m
1↑m             ⍝ the first row (still a table)
¯1↑m            ⍝ the first row from the end (still a table)
1↓m             ⍝ everything but the first row
```

#### 7. Flip a table: `⍉`

- `⍉` (transpose) swaps rows and columns. A 2×3 table becomes 3×2.

```
%%bpl
m
⍉m              ⍝ rows become columns
ρ⍉m             ⍝ 3 2
```

### Nesting

#### 8. A list of lists

- A list's items can themselves be lists. Write each inner list in its own brackets: `[[1 2] [3 4 5]]`.
- The output draws a box round each inner list.
- `≢x` counts the **outer** items: 2. `≢¨x` counts inside **each** one: 2 and 3. (`¨` means "each".)
- `+/¨x` adds up each inner list.

```
%%bpl
x←[[1 2] [3 4 5]]    ⍝ two lists inside one
x
≢x                   ⍝ 2 items
≢¨x                  ⍝ their lengths: 2 3
+/¨x                 ⍝ their totals: 3 12
```

#### 9. A list of words is nested too

- `"the" "cat" "sat"` is a list of three items, and each item is a list of characters.
- `⌽s` reverses the **order of the words**. `⌽¨s` reverses **each word**.

```
%%bpl
s←"the" "cat" "sat"
s
≢s              ⍝ 3 words
≢¨s             ⍝ 3 letters each
⌽s              ⍝ the words in reverse order
⌽¨s             ⍝ each word spelled backwards
⌽¨⌽s            ⍝ words in reverse order, and each word spelled backwards
```

#### 10. How deep: `≡`

- With one argument, `≡` gives the **depth**: how many levels of lists there are.
- A single number has depth 0, a plain list 1, and a list of lists 2.

```
%%bpl
x
≡5              ⍝ 0: just a number
≡1 2 3          ⍝ 1: a plain list
≡x              ⍝ 2: a list of lists
```

#### 11. Wrap it up as one item: `⊂`

- `⊂` (**enclose**) puts a box around a whole array. The box counts as **one item**, no matter how many things are inside it.
- Example: `1 2 3` has 3 items, but `⊂1 2 3` is one box containing that list. So `≢⊂1 2 3` is 1.
- This matters when joining words. `s,"mat"` opens the word and adds its 3 letters separately. `s,⊂"mat"` keeps `"mat"` boxed as one word.
- Result: `s,"mat"` has 6 items (3 words + 3 letters), while `s,⊂"mat"` has 4 items (4 words).

```
%%bpl
≢1 2 3          ⍝ 3 items
≢⊂1 2 3         ⍝ 1: one wrapped item
s
s,"mat"         ⍝ adds 3 letters (6 items)
s,⊂"mat"        ⍝ adds 1 word (4 items)
```

**Why does `s,"mat"` add three letters?**

Comma has two different jobs:

- **Monadic** `,x` ravels (flattens) `x`.
- **Dyadic** `a,b` joins `a` and `b` along the last axis.

So dyadic `,` does **not** first ravel everything. In `s,"mat"`, the right argument `"mat"` is already a list of three characters, so those characters are joined to `s` as three separate items.

Use `s,⊂"mat"` to enclose `"mat"` first, making the whole word one item before it is joined.

#### 12. Pick an item, and turn lists into a table

- `n⊃x` (**pick**) selects item `n`, counting from 0, and removes that item's outer box. It unwraps **one selected item**; it does not flatten the contents.
- Here `x` is a list of lists, so `1⊃x` selects the second boxed list and returns the plain list `3 4 5`.
- For a plain list such as `v`, its items are numbers rather than boxed lists. Therefore `1⊃v` simply returns the second number, `20`.
- With one argument, `⊃` (**mix**) turns a list of **equal-length** lists into a table: each inner list becomes a row.

```
%%bpl
x
1⊃x                  ⍝ nested list: pick the second inner list and unwrap, 3 4 5
v
1⊃v                  ⍝ plain list: pick the second item, 20
⊃[[1 2] [3 4]]       ⍝ mix: two equal-length lists become two rows
```
