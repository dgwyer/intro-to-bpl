## Getting data in and out

So far all our data has been typed into the notebook. This dialog shows how to **save data to files and read it back**, as plain text, CSV and JSON. Then it shows how BPL and **Python** pass data to each other in the same notebook.

The dialog writes a few small example files into a folder called `bpl-files`, and the last step deletes it again.

BPL docs: [Data and files](https://answerdotai.github.io/basedpl/data.html) · [Python](https://answerdotai.github.io/basedpl/python.html) · [Notebooks](https://answerdotai.github.io/basedpl/notebooks.html)

Glyphs and ideas this dialog teaches:

| Glyph | Plain English |
|---|---|
| `•mkdir "dir"` | make a folder |
| `o •nput text` | write text to a file. `o` is a record of options: the `path`, and `"overwrite":1` to replace an old file |
| `•nget "file"` | read a text file |
| `•csv⁻ table` | turn a table into CSV text |
| `•csv text` | turn CSV text into a table |
| `•json⁻ x` | turn a record or list into JSON text |
| `•json text` | turn JSON text into records and lists |
| `•vfi text` | find the numbers in a piece of text |
| `•readdir "dir"` | list a folder's files, as a table |
| `["recurse":1] •remove "dir"` | delete a folder and everything in it |
| `bpl['name'] = …` | (Python) store a Python value in BPL |
| `x = %bpl …` | (Python) run BPL and keep the answer in Python |
| `.py` `.np` | (Python) turn a BPL answer into Python values or a NumPy array |
| `fn('name')` | (Python) use a BPL function from Python |

### Text files

#### 1. Where files go

- A file name like `"bpl-files/hello.txt"` is found from the notebook's current folder. This Python cell shows which folder that is.
- `•mkdir` makes a folder (and any missing parents). It gives nothing back on screen.

```
import os
os.getcwd()          # the folder that file names start from
```

```
%%bpl
•mkdir "bpl-files"   ⍝ a folder for this dialog's files
```

#### 2. Write a file, then read it

- `•nput` writes text to a file. Its options go in a record on the left: `path` is the file name.
- `"overwrite":1` lets it replace a file that's already there. Without it, writing the same file a second time is an error, so a cell couldn't be run twice.
- `•nget` reads the file back as text.

```
%%bpl
o←["path":"bpl-files/hello.txt" "overwrite":1]
o •nput "Hello from BPL"          ⍝ write
•nget "bpl-files/hello.txt"       ⍝ read it back
```

#### 3. Several lines

- A file with several lines is one piece of text with a **newline** character (`•ucs 10`) between them.
- To write a list of lines: `lines,¨⊂nl` adds a newline to the end of each, and `∊` flattens them into one text.
- To read lines back: `(t≠nl)⊆t` cuts the text wherever there's a newline.

```
%%bpl
nl←•ucs 10                          ⍝ newline
lines←"Mon 12" "Tue 15" "Wed 11"
text←∊lines,¨⊂nl                    ⍝ one text, a newline after each line
["path":"bpl-files/days.txt" "overwrite":1] •nput text
t←•nget "bpl-files/days.txt"
t
(t≠nl)⊆t                            ⍝ back to three lines
```

### Tables and records in files

#### 4. CSV: a table in a file

- CSV is a text format for tables, which spreadsheets can open.
- `•csv⁻ week` turns a table into CSV text. The `⁻` means "the other way round": `•csv` reads CSV, and `•csv⁻` writes it.
- `•csv` of the file's text gives the table back. `≡` checks it's the same table.
- In the file, the temperatures are written as `12.0`. Plain numbers in BPL are approximate (decimals), and the file says so.

```
%%bpl
days←"Mon" "Tue" "Wed"
degrees←12 15 11
week←["day":days "temp":degrees]   ⍝ a small table
csv←•csv⁻ week                     ⍝ table → CSV text
csv
["path":"bpl-files/week.csv" "overwrite":1] •nput csv
w←•csv •nget "bpl-files/week.csv"   ⍝ file → table
w
w≡week                              ⍝ $t: the same table
```

#### 5. JSON: records and lists in a file

- JSON is a text format for records and lists, used by most web services.
- `•json⁻` turns a record into JSON text, and `•json` reads it back.
- In the output, each `"` inside the text is shown doubled (`""`). That's how BPL displays a quote mark inside text. The file itself has single quote marks.

```
%%bpl
tricks←"sit" "roll"
pet←["name":"Rex" "age":3 "tricks":tricks]
j←•json⁻ pet                       ⍝ record → JSON text
j
["path":"bpl-files/pet.json" "overwrite":1] •nput j
p←•json •nget "bpl-files/pet.json"  ⍝ file → record
p
p.tricks
```

#### 6. Numbers from text: `•vfi`

- `•vfi` reads the numbers in a text. It gives two lists: `1` or `0` for "was it a valid number?", and the numbers themselves (0 where it wasn't).
- `↑v` is the first list and `1⊃v` the second, so `(↑v)#1⊃v` keeps just the real numbers.

```
%%bpl
v←•vfi "12 15 abc 11"
v                   ⍝ valid flags, then the numbers
(↑v)#1⊃v            ⍝ just the valid numbers: 12 15 11
(↑v)#↑¯1↑v          ⍝ alternative method
```

#### 7. What's in the folder?

- `•readdir` lists a folder as a table, with one row per file. Its columns include `name` and `len` (the size in bytes).

```
%%bpl
r←•readdir "bpl-files"
r.name              ⍝ the file names
r.len               ⍝ their sizes in bytes
```

### BPL and Python

`%%bpl` cells and ordinary Python cells share **one** BPL workspace. Python reaches it through `bpl`, imported from `basedpl`.

#### 8. From Python into BPL

- In a Python cell, `bpl['temps'] = […]` stores a Python list in BPL under the name `temps`.
- A `%%bpl` cell can then use `temps` like any other name.

```
from basedpl import bpl, fn
bpl['temps'] = [12, 15, 11, 18, 20, 14, 9]    # Python list → BPL name
```

```
%%bpl
temps
+/temps             ⍝ BPL sees the Python data: 99
```

#### 9. From BPL back to Python

- In a Python cell, `x = %bpl …` (one `%`) runs one line of BPL and keeps the answer.
- `.np` turns it into a NumPy array, and `.py` into plain Python values. A record becomes a Python dictionary.

```
t = %bpl temps>14
t.np                 # a NumPy array of True/False
```

```
w = %bpl week
w.py                 # the table as a Python dictionary
```

#### 10. Use a BPL function from Python

- Define a function in BPL, then `fn('mean')` gives Python a function that runs it.
- Python whole numbers become BPL whole numbers (`ₓ`), and `÷` on whole numbers rounds down. So the mean of `[1, 2, 4]` comes back as 2.
- Pass at least one decimal, such as `1.0`, to get the decimal mean, 2.33…

```
%%bpl
mean←+/÷≢
```

```
mean = fn('mean')
mean([1, 2, 4]).py, mean([1.0, 2, 4]).py   # whole numbers: 2; decimals: 2.33…
```

- `.py` is needed because `fn('mean')` gives back a BPL value (a `basedpl.Array`), not a Python one. BPL functions take and return BPL arrays.
- `.py` turns that BPL value into plain Python, here an `int` (`2`) or a `float` (`2.33…`). Without it you'd still have the BPL array, and Python can't use it like an ordinary number or list.
- `.np` does the same job but gives a NumPy array.

### Tidy up

#### 11. Delete the example folder

- `•remove` deletes a file or an empty folder. With `["recurse":1]` it deletes a folder **and everything in it**, so use it with care.
- Run the cells above again any time to make the files again.

```
%%bpl
["recurse":1] •remove "bpl-files"   ⍝ delete the folder and its files
```
