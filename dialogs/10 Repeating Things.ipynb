## Repeating a step

Many calculations are "do this step again and again": interest added each year, a guess improved until it's good enough, a running total, a random walk. In BPL you rarely write a loop. Instead, `⍣` repeats a function, and `\` (scan) keeps a running result along a list.

BPL docs: [Power `⍣`](https://answerdotai.github.io/basedpl/glyphs/star-diaeresis.html) · [Scan `\`](https://answerdotai.github.io/basedpl/glyphs/backslash.html) · [Roll `¿`](https://answerdotai.github.io/basedpl/glyphs/inverted-question.html)

Glyphs and ideas this dialog teaches:

| Glyph | Plain English |
|---|---|
| `f⍣3` | power: do `f` three times |
| `f⍣(ιn)` | do `f` again and again, keeping every step (`n` values, starting with the start value) |
| `f⍣{α≥200}` | keep doing `f` until the test is true (`α` is the new value) |
| `f⍣≡` | keep doing `f` until the answer stops changing |
| `f⍣[test]` | the same, keeping every step |
| `f\` | scan: a running result along a list |
| `s f\` | a scan that starts from `s` |
| `¿n` | roll: a random whole number from `0` to `n-1` |
| `•prefs ["seed":42]` | fix the random numbers, so you get the same ones each time |

Other glyphs used along the way:

| Glyph | Plain English |
|---|---|
| `←` | store a value or function in a name |
| `⍝` | comment: BPL ignores the rest of the line |
| `2×` | a function: "times 2" (a value fixed on the left) |
| `{ }` and `ω` | a function written with braces; `ω` is its right argument |
| `⁝` | keeps two things apart: `f⍣3⁝1` keeps the count `3` separate from the start value `1` |
| `+ - × ÷` | add, subtract, multiply, divide |
| `ι5` | the positions `0 1 2 3 4` |
| `≢` | count the items |
| `+/` | total of a list |
| `√` | square root |
| `⌈` `⌊` | the bigger / smaller of two numbers |
| `=` `≥` | equal / at least (answers `$t` or `$f`) |
| `10ρ6` | reshape: ten copies of 6 |
| `¯` | minus sign for a negative number |
| `ₓ` | (in output) marks an exact number |

### A set number of times

#### 1. Repeat a step: `⍣`

- `double←2×` is "times 2".
- `double⍣3` means "double, three times".
- In the code below `⁝` keeps the count `3` apart from the start value `1`. Without it, `3 1` would join into one list.

```
%%bpl
double←2×        ⍝ times 2
double⍣3⁝1       ⍝ 1 → 2 → 4 → 8
```

BPL parses `double⍣3⁝1` as “the function `double`, repeated 3 times, applied to the starting input `1`.” The first call uses `1`; every later call uses the previous result:

`1 → double(1)=2 → double(2)=4 → double(4)=8`

This resembles recursion because each step depends on the previous result, but it is iteration (repeated function composition): `double(double(double(1)))`.

#### 2. Savings with interest

- 5% interest a year means "times 1.05" each year: `grow←1.05×`.
- `grow⍣10⁝100` is £100 after 10 years.

```
%%bpl
grow←1.05×       ⍝ add 5%
grow⍣10⁝100      ⍝ £100 after 10 years
```

#### 3. Keep every step

- Put a list of counts in brackets: `⍣(ι5)` gives the value after 0, 1, 2, 3 and 4 steps.
- So the answer starts with the start value itself.

```
%%bpl
double⍣(ι5)⁝1    ⍝ 1 2 4 8 16
grow⍣(ι4)⁝100    ⍝ the balance in years 0, 1, 2 and 3
```

### Until something happens

#### 4. Repeat until a test is true

- Put a test in braces after `⍣`. Inside it, `α` is the **new** value (and `ω` the one before).
- `grow⍣{α≥200}⁝100` keeps adding interest until the balance reaches £200.
- To see how long that took, keep every step with `⍣[{α≥200}]` (the test in square brackets), then count them. The list includes year 0, so the number of years is one less than its length.

```
%%bpl
grow⍣{α≥200}⁝100          ⍝ the first balance of at least £200
h←grow⍣[{α≥200}]⁝100      ⍝ every year's balance up to then
¯1+≢h                     ⍝ 15 years
```

#### 5. Repeat until the answer stops changing: `⍣≡`

- A classic way to find √2: start with a guess, then replace it with the average of the guess and 2÷guess. Each step gets closer.
- `⍣≡` repeats until a step gives the same answer as the one before.
- `⍣[≡]` keeps every guess, so you can watch it settle.

```
%%bpl
step←{0.5×ω+2÷ω}    ⍝ average of ω and 2÷ω
step⍣≡⁝1            ⍝ √2
step⍣[≡]⁝1          ⍝ every guess along the way
```

### Running results: scans

#### 6. A running total: `+\`

- `+\` adds as it goes: the first item, then the first two, then the first three, and so on.
- Here: rain for each day of a week (mm), and the total so far.

```
%%bpl
rain←0 2 5 0 0 1 8      ⍝ Mon to Sun
+\rain                  ⍝ total rain so far
+/rain                  ⍝ just the sum
```

#### 7. Other running results

- Any function that takes two arguments can scan. `⌈` gives the bigger of two numbers, so `⌈\` is "the biggest so far".

```
%%bpl
temps←12 15 11 18 20 14 9   ⍝ Mon to Sun
⌈\temps                    ⍝ hottest so far
```

#### 8. A scan that starts somewhere

- A value on the left is the starting point: `100+\` starts the running total at 100.
- Here: a balance of £100, then deposits of 10 and 20 and a withdrawal of 5.

```
%%bpl
100+\10 20 ¯5       ⍝ 110 130 125
```

### Simulations

#### 9. Random numbers: `¿`

- `¿6` gives a random whole number from 0 to 5. So `1+¿6` is a die roll.
- `10ρ6` is ten 6s, so `1+¿10ρ6` is ten die rolls.
- `•prefs ["seed":42]` fixes the random numbers, so you see the same rolls as this dialog. Change 42 to get others.

```
%%bpl
•prefs ["seed":42]   ⍝ repeatable random numbers
1+¿10ρ6              ⍝ ten die rolls
```

#### 10. A random walk with a scan

- Each step is `¯1` or `1`: `¿20ρ2` gives twenty 0s and 1s, and `¯1+2×` turns 0 into ¯1 and 1 into 1.
- The position after each step is the running total of the steps: `+\steps`.

```
%%bpl
•prefs ["seed":7]
steps←¯1+2×¿20ρ2     ⍝ twenty steps of ¯1 or 1
steps
+\steps              ⍝ position after each step
+/steps              ⍝ normal sum
```

#### 11. The same walk with `⍣`

- Instead, write **one step** as a function, `walk←{ω+¯1+2×¿2}`, and repeat it, keeping every position.
- The seed makes this walk repeatable too. It is a different path from step 10, though: it draws one random number per step instead of twenty at once. The list also includes the start position 0.
- Scan suits it when you can make all the steps first. `⍣` suits it when each step depends on where you are.

```
%%bpl
•prefs ["seed":7]
walk←{ω+¯1+2×¿2}     ⍝ one step from position ω
walk⍣(ι21)⁝0         ⍝ start at 0, keep 21 positions
```
